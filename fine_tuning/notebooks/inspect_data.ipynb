# Inspect Finetuning Data

```python
import xarray as xr
import numpy as np
import glob
```

## 2D Data files: Surface Variables

```python

# Step 1: Load and examine 2D surface file
print("="*50)
print("EXAMINING 2D SURFACE FILE")
print("="*50)

files_2d = sorted(glob.glob('/home/user/Documents/aurora/data_out/2d/*.nc'))
if files_2d:
    file_path = files_2d[0]
    ds_2d = xr.open_dataset(file_path)
    
    print(f"File: {file_path}")
    print(f"\nDimensions: {ds_2d.dims}")
    
    print("\nCoordinates:")
    for name, var in ds_2d.coords.items():
        print(f"- {name}: {var.shape} {var.dtype}")
    
    print("\nVariables in 2D file:")
    for name, var in ds_2d.data_vars.items():
        shape_str = str(var.shape)
        size_mb = var.nbytes / (1024 * 1024) if hasattr(var, 'nbytes') else "N/A"
        print(f"- {name}: {shape_str} {var.dtype} ({size_mb:.2f} MB)")
    
    ds_2d.close()
else:
    print("No 2D files found")
    ds_2d = None

# Step 2: Load and examine 3D atmospheric file
print("\n" + "="*50)
print("EXAMINING 3D ATMOSPHERIC FILE")
print("="*50)

```

## 3D Data Files: Atmospheric Variables

```python
files_3d = sorted(glob.glob('/home/user/Documents/aurora/data_out/3d/*.nc'))
if files_3d:
    file_path = files_3d[0]
    ds_3d = xr.open_dataset(file_path)
    
    print(f"File: {file_path}")
    print(f"\nDimensions: {ds_3d.dims}")
    
    print("\nCoordinates:")
    for name, var in ds_3d.coords.items():
        print(f"- {name}: {var.shape} {var.dtype}")
    
    print("\nVariables in 3D file:")
    for name, var in ds_3d.data_vars.items():
        shape_str = str(var.shape)
        size_mb = var.nbytes / (1024 * 1024) if hasattr(var, 'nbytes') else "N/A"
        print(f"- {name}: {shape_str} {var.dtype} ({size_mb:.2f} MB)")
    
    ds_3d.close()
else:
    print("No 3D files found")
    ds_3d = None

# Step 3: Create Aurora to WRF variable mapping
print("\n" + "="*50)
print("AURORA TO WRF VARIABLE MAPPING")
print("="*50)

```

```python
aurora_to_wrf = {
    # Static variables
    'z': ['HGT', 'TERRAIN', 'Z'],         # Geopotential height or terrain height
    'lsm': ['LANDMASK', 'LAND'],          # Land-sea mask (1=land, 0=water)
    'slt': ['SOILCAT', 'ISLTYP', 'LU_INDEX'], # Soil type or land use category
    
    # Surface variables
    't2m': ['T2'],                         # 2-meter temperature
    'u10': ['U10'],                        # 10-meter u-wind
    'v10': ['V10'],                        # 10-meter v-wind
    'msl': ['PSFC', 'SLP', 'PMSL'],        # Surface or sea level pressure
    
    # Atmospheric variables
    't': ['T', 'TK'],                      # Air temperature
    'u': ['U'],                            # U wind component
    'v': ['V'],                            # V wind component
    'q': ['QVAPOR', 'Q']                   # Specific humidity or water vapor 
}

# Step 4: Check variable availability in each file type
print("\nChecking for required Aurora variables:")
print("-"*40)

for aurora_var, wrf_vars in aurora_to_wrf.items():
    print(f"\nAurora variable: {aurora_var}")
    found = False
    
    # Check in 2D file
    if files_2d:
        ds_2d = xr.open_dataset(files_2d[0])
        for var in wrf_vars:
            if var in ds_2d.variables:
                shape_str = str(ds_2d[var].shape)
                found = True
                print(f"  ✓ Found as {var} in 2D file: {shape_str}")
        ds_2d.close()
    
    # Check in 3D file
    if files_3d:
        ds_3d = xr.open_dataset(files_3d[0])
        for var in wrf_vars:
            if var in ds_3d.variables:
                shape_str = str(ds_3d[var].shape)
                found = True
                print(f"  ✓ Found as {var} in 3D file: {shape_str}")
        ds_3d.close()
    
    if not found:
        print(f"  ✗ Not found in available files")

print("\n" + "="*50)
print("VARIABLE MAPPING SUMMARY")
print("="*50)

```

```python
# Print a clean summary of the mapping we'll use
if files_2d or files_3d:
    found_mappings = {}
    
    # Define variable types
    surface_static_vars = ['z', 'lsm', 'slt', 't2m', 'u10', 'v10', 'msl']
    atmospheric_vars = ['t', 'u', 'v', 'q']
    
    # Check 2D file for surface and static variables
    if files_2d:
        ds_2d = xr.open_dataset(files_2d[0])
        for aurora_var, wrf_vars in aurora_to_wrf.items():
            if aurora_var in surface_static_vars:  # Only check surface/static variables
                for var in wrf_vars:
                    if var in ds_2d.variables and aurora_var not in found_mappings:
                        found_mappings[aurora_var] = {'wrf_var': var, 'source': '2D file'}
        ds_2d.close()
    
    # Check 3D file for ALL variables, with priority for atmospheric variables
    if files_3d:
        ds_3d = xr.open_dataset(files_3d[0])
        
        # First check atmospheric variables (prioritizing 3D file)
        for aurora_var in atmospheric_vars:
            if aurora_var in aurora_to_wrf:
                for var in aurora_to_wrf[aurora_var]:
                    if var in ds_3d.variables:
                        found_mappings[aurora_var] = {'wrf_var': var, 'source': '3D file'}
                        break
        
        # Then check any remaining surface/static variables not found in 2D file
        for aurora_var in surface_static_vars:
            if aurora_var not in found_mappings and aurora_var in aurora_to_wrf:
                for var in aurora_to_wrf[aurora_var]:
                    if var in ds_3d.variables:
                        found_mappings[aurora_var] = {'wrf_var': var, 'source': '3D file'}
                        break
        ds_3d.close()
    
    print("\nProposed variable mapping for Aurora:")
    print("\nSurface and Static variables:")
    for aurora_var in surface_static_vars:
        if aurora_var in found_mappings:
            print(f"  {aurora_var} ← {found_mappings[aurora_var]['wrf_var']} ({found_mappings[aurora_var]['source']})")
        else:
            print(f"  {aurora_var} ← Missing (needs derivation or external source)")
            
    print("\nAtmospheric variables (3D):")
    for aurora_var in atmospheric_vars:
        if aurora_var in found_mappings:
            print(f"  {aurora_var} ← {found_mappings[aurora_var]['wrf_var']} ({found_mappings[aurora_var]['source']})")
        else:
            print(f"  {aurora_var} ← Missing (needs derivation or external source)")
```
