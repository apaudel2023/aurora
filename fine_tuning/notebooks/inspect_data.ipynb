# Inspect Finetuning Data

```python
import glob

import xarray as xr
```

## 2D Data files: Surface Variables

```python
# Step 1: Load and examine 2D surface file
print("=" * 50)
print("EXAMINING 2D SURFACE FILE")
print("=" * 50)

files_2d = sorted(glob.glob("/home/user/Documents/aurora/data_out/2d/*.nc"))
if files_2d:
    file_path = files_2d[0]
    ds_2d = xr.open_dataset(file_path)

    print(f"File: {file_path}")
    print(f"\nDimensions: {ds_2d.dims}")

    print("\nCoordinates:")
    for name, var in ds_2d.coords.items():
        print(f"- {name}: {var.shape} {var.dtype}")

    print("\nVariables in 2D file:")
    for name, var in ds_2d.data_vars.items():
        shape_str = str(var.shape)
        size_mb = var.nbytes / (1024 * 1024) if hasattr(var, "nbytes") else "N/A"
        print(f"- {name}: {shape_str} {var.dtype} ({size_mb:.2f} MB)")

    ds_2d.close()
else:
    print("No 2D files found")
    ds_2d = None

# Step 2: Load and examine 3D atmospheric file
print("\n" + "=" * 50)
print("EXAMINING 3D ATMOSPHERIC FILE")
print("=" * 50)
```

## 3D Data Files: Atmospheric Variables

```python
files_3d = sorted(glob.glob("/home/user/Documents/aurora/data_out/3d/*.nc"))
if files_3d:
    file_path = files_3d[0]
    ds_3d = xr.open_dataset(file_path)

    print(f"File: {file_path}")
    print(f"\nDimensions: {ds_3d.dims}")

    print("\nCoordinates:")
    for name, var in ds_3d.coords.items():
        print(f"- {name}: {var.shape} {var.dtype}")

    print("\nVariables in 3D file:")
    for name, var in ds_3d.data_vars.items():
        shape_str = str(var.shape)
        size_mb = var.nbytes / (1024 * 1024) if hasattr(var, "nbytes") else "N/A"
        print(f"- {name}: {shape_str} {var.dtype} ({size_mb:.2f} MB)")

    ds_3d.close()
else:
    print("No 3D files found")
    ds_3d = None

# Step 3: Create Aurora to WRF variable mapping
print("\n" + "=" * 50)
print("AURORA TO WRF VARIABLE MAPPING")
print("=" * 50)
```

```python
# Define the mapping dictionary
aurora_to_wrf = {
    # Static variables
    "z": ["HGT", "TERRAIN", "Z"],  # Geopotential height or terrain height
    "lsm": ["LANDMASK", "LAND"],  # Land-sea mask (1=land, 0=water)
    "slt": ["SOILCAT", "ISLTYP", "LU_INDEX"],  # Soil type or land use category
    # Surface variables
    "t2m": ["T2"],  # 2-meter temperature
    "u10": ["U10"],  # 10-meter u-wind
    "v10": ["V10"],  # 10-meter v-wind
    "msl": ["PSFC", "SLP", "PMSL"],  # Surface or sea level pressure
    # Atmospheric variables
    "t": ["T", "TK"],  # Air temperature
    "u": ["U"],  # U wind component
    "v": ["V"],  # V wind component
    "q": ["QVAPOR", "Q"],  # Specific humidity or water vapor
}

# Find files
files_2d = sorted(glob.glob("/home/user/Documents/aurora/data_out/2d/*.nc"))
files_3d = sorted(glob.glob("/home/user/Documents/aurora/data_out/3d/*.nc"))

# Print summary of what's available and what's missing
print("\n" + "=" * 60)
print("SUMMARY OF VARIABLE AVAILABILITY")
print("=" * 60)

# Define variable categories
static_vars = ["z", "lsm", "slt"]
surface_vars = ["t2m", "u10", "v10", "msl"]
atmospheric_vars = ["t", "u", "v", "q"]

found_vars = {}
missing_vars = []

# First get all variables and their shapes in each file
vars_2d = {}
vars_3d = {}

if files_2d:
    ds = xr.open_dataset(files_2d[0])
    for var in ds.variables:
        vars_2d[var] = ds[var].shape
    ds.close()

if files_3d:
    ds = xr.open_dataset(files_3d[0])
    for var in ds.variables:
        vars_3d[var] = ds[var].shape
    ds.close()

# Check for each Aurora variable
for aurora_var, wrf_vars in aurora_to_wrf.items():
    var_found = False
    sources_with_shapes = []
    found_wrf_var = None

    # Try to find in both file types
    for wrf_var in wrf_vars:
        if wrf_var in vars_2d:
            sources_with_shapes.append(f"2D file (shape: {vars_2d[wrf_var]})")
            found_wrf_var = wrf_var
            var_found = True

        if wrf_var in vars_3d:
            sources_with_shapes.append(f"3D file (shape: {vars_3d[wrf_var]})")
            found_wrf_var = wrf_var
            var_found = True

        if var_found:
            found_vars[aurora_var] = {
                "wrf_var": found_wrf_var,
                "sources_with_shapes": sources_with_shapes,
            }
            break

    if not var_found:
        missing_vars.append(aurora_var)

# Print final availability summary by category
print("\n1. STATIC VARIABLES:")
print("-" * 40)
for var in static_vars:
    if var in found_vars:
        sources_str = " and ".join(found_vars[var]["sources_with_shapes"])
        print(f"  ✓ {var} ← {found_vars[var]['wrf_var']} ({sources_str})")
    else:
        print(f"  ✗ {var} (missing)")

print("\n2. SURFACE VARIABLES:")
print("-" * 40)
for var in surface_vars:
    if var in found_vars:
        sources_str = " and ".join(found_vars[var]["sources_with_shapes"])
        print(f"  ✓ {var} ← {found_vars[var]['wrf_var']} ({sources_str})")
    else:
        print(f"  ✗ {var} (missing)")

print("\n3. ATMOSPHERIC VARIABLES:")
print("-" * 40)
for var in atmospheric_vars:
    if var in found_vars:
        sources_str = " and ".join(found_vars[var]["sources_with_shapes"])
        print(f"  ✓ {var} ← {found_vars[var]['wrf_var']} ({sources_str})")
    else:
        print(f"  ✗ {var} (missing)")

# Print summary of missing variables
if missing_vars:
    print("\nSUMMARY OF MISSING VARIABLES:")
    print("-" * 40)
    for var in missing_vars:
        print(f"  ✗ {var} (will need to be derived or found from external source)")
```
